```python
#Generator function>>

#Regular function >> takes a list and give square of each of the list

def square_number(n):
    result = []
    for i in n:
        result.append(i**2)
    return result
square_number([1,2,3,10,12,13]) 
```

```python
#Regular function calculate the square of the values and return in one go
#Since its returning in one go>> memory usage will be more and it will take a lot of exceution time if there is lot of computation as all the computations will be done first and then the calculation will be returned in one go

def square_number(n):
    result = []
    for i in range(n):
        result.append(i**2)
    return result
square_number(10)
```

```python
#Can we have something to generate result one by one instead of one go.
#yield>> general function uses return statement but a generator function use yield statement

def square_number_gene(n):
    for i in range(n):
        yield i ** 2
square_number_gene(10)
```

```python
gen = square_number_gene(10)
gen 
```

```python
next(gen)
```

```python
#fibonacci series>> 0,1,1,2,3,5,8.......

def fib(n):
    a = 0
    b = 1
    for i in range (n):
        yield a
        a,b = b, a+b
```

```python
f = fib(10000)
f
```

```python
next(f)
```
