```python
#csv>>comma separated files
data = [["Name","Course","Fee"],
        ["Sam","Python", "300000"],
        ["Deni","DS","400000"]]
```

```python
for i in data:
    print(i)
```

```python
import csv
with open("example_csv.csv","w") as f:
    w = csv.writer(f)
    for i in data:
        w.writerow(i)
```

```python
with open("example_csv.csv","r") as f:
    r = csv.reader(f)
    for i in r:
        print(i)
```

```python
#Buffered reading and writing 
#it allows the program to read and write data in chunks 
```

```python
import io
with open('test_buf.txt',"wb") as f:
    file = io.BufferedWriter(f)
    file.write(b"A computer is a machine that can be programmed to carry out sequences of arithmetic or logical operations automatically. Modern computers can perform generic sets of operations known as programs. These programs enable computers to perform a wide range of tasks.The term computer may refer to a normally complete computer that includes the hardware, the operating system, software and peripheral equipment needed and used for full operation; or to a group of computers that are linked and function together, such as a computer network or computer cluster.")

    file.write(b"Hello World, how are u?")
    file.write(b"I am Studying DS")
    file.flush()
```

```python
with open('test_buf.txt',"rb") as f:
    file = io.BufferedReader(f)
    data = file.read(150)   #read the first chunk of 100 bytes
    print(data)
```
