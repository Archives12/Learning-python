```python
text_content = """
example1
example2
example3
"""
```

```python
with open("example.txt", "w") as file:
    file.write(text_content)
```

```python
with open("example.txt", "r") as file:
    data = file.read()
print(data)
```

```python
#csv>>comma separated files
data = [["Name","Course","Fee"],
        ["Sam","Python", "300000"],
        ["Deni","DS","400000"]]
```

```python
for i in data:
    print(i)
```

```python
import csv
with open("example_csv.csv","w") as f:
    w = csv.writer(f)
    for i in data:
        w.writerow(i)
```

```python
with open("example_csv.csv","r") as f:
    r = csv.reader(f)
    for i in r:
        print(i)
```
