```python
#Extract odd and even numbers from a list
class ListOps:
    l = [1,2,3,4,5]
    def extract_even(self,l):
        l1 = []
        for i in l:
            if i%2==0:
                l1.append(i)
        return l1
    def extract_odd(self,l):
        l2 = []
        for i in l:
            if i%2!=0:
                l2.append(i)
        return l2 
```

```python
ops1 = ListOps()
ops1.l
```

```python
ops1.extract_even(ops1.l)
```

```python
ops1.extract_odd(ops1.l)
```

```python
#you want to take the attribute value of the class for each object differently

class ListOps:
    l = [1,2,3,4,5]
    def __init__(self, l): 
        self.l = l
    def extract_even(self,l):
        l1 = []
        for i in l:
            if i % 2 == 0:
                l1.append(i)
        return l1
    def extract_odd(self,l):
        l2 = []
        for i in l:
            if i % 2 != 0:
                l2.append(i)
        return l2 
```

```python
class ListOps:
    l = [1,2,3,4,5]
    def __init__(self, l):      #magic method/ dunder method. init means initialisation>> of variable
        self.l2 = l         #Everywhere you will be able to access using self.l2
    def extract_even(self,):    #be careful as init has been used only pass self in the method
        l = self.l2
        l1 = []
        for i in l:
            if i % 2 == 0:
                l1.append(i)
        return l1
    def extract_odd(self,l):
        l = self.l2
        l1 = []
        for i in l:
            if i % 2 != 0:
                l1.append(i)
        return l1 
```
