```python
def only_numeric(a):
    n = []
    for i in a:
        if type(i) == list:
            for j in i:
                if type(j) == int or type(j) == float:
                    n.append(j)
        else:
            if type(i) == int or type(i) == float:
                n.append(i)
    return n
```

```python
only_numeric([1,2.2,"Ajay", True, (3+7j),"pwksills",5,6,[100,101,"Ajay"]])
```

```python
#Write a function can take any number of values and return the list

def return_list(*args):
    l = []
    for i in args:
        if type(i) == list:
            l.append(i)
    return l
```

```python
return_list(1,2,2.3,5+7j, {2,3}, [1,2,3,4] , (2,3), True,"Ajay")
```

```python
#A function which can take any number of key value pair and return only list key values arguement

def test(**kwargs):
    for i in kwargs.keys():
        if type(kwargs[i]) == list:
            return i, kwargs[i]
```

```python
test (a=2, b=3, c=[1,2,3], d={1,2}, e=(1,2))
```

```python
#function inside function

def marks_in_subject(**kwargs):
    marks_list = []
    for sub, marks in kwargs.items():
        marks_list.append(marks)
    return marks_list
marks_in_subject(math = 90, science = 95, english = 85)
```

```python
def marks_in_subject(**kwargs):
    def total_marks(marks_list):
        return sum(marks_list)
    marks_list = [] 
    for sub, marks in kwargs.items():
        marks_list.append(marks)
    return total_marks(marks_list)

marks_in_subject(math = 90, science = 95, english = 85)
```

```python
#Calling function from outside

def total_marks(marks_list):
    return sum(marks_list)

def marks_in_subject(**kwargs):
    marks_list = [] 
    for sub, marks in kwargs.items():
        marks_list.append(marks)
    return total_marks(marks_list)

marks_in_subject(math = 90, science = 95, english = 85)
```

```python
#Documentation string >> docstring >> it is a string which is used to describe the function, its arguement and its output
def find_power(num, power):
    """This function returns the power of number
    Args: 
        num(int) : provide any integer as input
        power(int): provide power as integer
    Returns:
        integer:Power of number
    """
    return num ** power
```
