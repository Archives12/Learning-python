```python
#Factorial of 5 >> 5*4*3*2*1 and 0! = 1

def fact_num(n):
    if n == 0:
        return 1
    else:
        return n*fact_num(n-1)
fact_num(5)
```

```python
fact_num = lambda n : 1 if n==0 else n * fact_num(n-1)
fact_num(5)
```

```python
#map>> executes a specified function for each of item of an iterable
#syntax >> map(func, *iterables)

l = [1,2,3,4,5]
def square(l):
    sq = []
    for i in l:
        sq.append(i**2)
    return sq

square(l)
```

```python
#reduce >> folding / reduction
#syntax >> #reduce(func, iterable)
#reduce will always take two argument function from functools import reduce

from functools import reduce
l = [2,1,3,4,4,5,6]
reduce(lambda x,y: x+y,l)
```

```python
reduce(lambda x,y,z: x+y+z,l)
```

```python
reduce(lambda x,y : x*y,l)
```

```python
words = ["Data", "Science","Course"]
reduce(lambda x,y: x + " " + y, words)
```

```python
#Highest value>> at each stage you are comparing x and y
numbers = [1,2,5,3,100,1000,10000]
reduce(lambda x,y : x if x>y else y, numbers)
```

```python
def factorial (n):
    return reduce(lambda x,y : x*y, range(1,n+1))
factorial(5)
```
